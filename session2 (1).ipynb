```python
#easy
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create circuit
qc = QuantumCircuit(1)

# Initial state is |0>

# Apply X gate
qc.x(0)

print("Quantum Circuit:")
print(qc)

# Statevector
state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

# Draw Circuit
qc.draw("mpl")
```

```python
#medium
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

qc = QuantumCircuit(1)

# Hadamard Gate
qc.h(0)

# Phase Gate
qc.s(0)

print(qc)

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

qc.draw("mpl")
plt.show()

plot_bloch_multivector(state)
plt.show()
```

```python
#hard
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_state_qsphere
import matplotlib.pyplot as plt

qc = QuantumCircuit(2)

qc.h(0)
qc.h(1)

print(qc)

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

qc.draw("mpl")
plt.show()

plot_state_qsphere(state)
plt.show()

print("\nProbabilities")

for s,v in state.to_dict().items():
    print(f"{s} : {abs(v)**2:.3f}")
```

```python
#real world
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

qc = QuantumCircuit(1)

qc.h(0)

# Random Phase Error
theta = np.random.uniform(0,0.3)

qc.p(theta,0)

print("Random Phase =",theta)

state = Statevector.from_instruction(qc)

print(state)

qc.draw("mpl")
plt.show()

plot_bloch_multivector(state)
plt.show()
```

```python
#challenge
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

def create_state(theta,phi):

    qc = QuantumCircuit(1)

    qc.ry(theta,0)
    qc.rz(phi,0)

    state = Statevector.from_instruction(qc)

    print("\nTheta =",theta)
    print("Phi =",phi)

    print(state)

    qc.draw("mpl")
    plt.show()

    plot_bloch_multivector(state)
    plt.show()


create_state(np.pi/4,np.pi/3)
```
