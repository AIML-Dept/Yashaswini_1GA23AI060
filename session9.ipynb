**Easy**

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)

qc.measure([0, 1], [0, 1])

print(qc.draw())

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

print("Output:", result.get_counts())
```

**Medium**

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def bell_state(state):
    qc = QuantumCircuit(2, 2)

    qc.h(0)
    qc.cx(0, 1)

    if state == "Phi-":
        qc.z(0)
    elif state == "Psi+":
        qc.x(0)
    elif state == "Psi-":
        qc.x(0)
        qc.z(0)

    qc.measure([0, 1], [0, 1])
    return qc

simulator = AerSimulator()

for state in ["Phi+", "Phi-", "Psi+", "Psi-"]:
    qc = bell_state(state)
    result = simulator.run(qc, shots=1024).result()

    print(state, ":", result.get_counts())
```

**Hard**

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)

# Hadamard basis measurement
qc.h(0)
qc.h(1)

qc.measure([0, 1], [0, 1])

print(qc.draw())

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

print("Output:", result.get_counts())
```

**Real-World**

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def qkd_demo(basis):
    qc = QuantumCircuit(2, 2)

    # Create an entangled pair
    qc.h(0)
    qc.cx(0, 1)

    # Select measurement basis
    if basis == "X":
        qc.h(0)
        qc.h(1)

    qc.measure([0, 1], [0, 1])

    result = AerSimulator().run(qc, shots=20).result()
    return result.get_counts()

print("Z basis:", qkd_demo("Z"))
print("X basis:", qkd_demo("X"))
```

**Challenge**

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import math

simulator = AerSimulator()

# GHZ state
ghz = QuantumCircuit(3, 3)
ghz.h(0)
ghz.cx(0, 1)
ghz.cx(1, 2)
ghz.measure([0, 1, 2], [0, 1, 2])

# W state
w = QuantumCircuit(3, 3)

w.ry(2 * math.acos(1 / math.sqrt(3)), 0)

# Replace controlled-H with a decomposed gate
w.ry(math.pi / 2, 1)
w.cx(0, 1)
w.ry(-math.pi / 2, 1)
w.cx(0, 1)

w.cx(1, 2)
w.cx(0, 1)
w.x(0)

w.measure([0, 1, 2], [0, 1, 2])

# Run GHZ simulation
ghz_result = simulator.run(
    ghz, shots=1024
).result()

print("GHZ state:", ghz_result.get_counts())

# Run W simulation
w_result = simulator.run(
    w, shots=1024
).result()

print("W state:", w_result.get_counts())
```
