```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import (
    plot_bloch_multivector,
    plot_state_qsphere,
    plot_histogram
)
from qiskit_aer import AerSimulator

import matplotlib.pyplot as plt

print("="*60)
print("Exercise 1 : X Gate")
print("="*60)

qc1 = QuantumCircuit(1)

print("\nInitial State = |0>")

qc1.x(0)

print("\nQuantum Circuit")
print(qc1)

qc1.draw("mpl")
plt.show()

state1 = Statevector.from_instruction(qc1)

print("\nStatevector")
print(state1)

print("\nBloch Sphere")
plot_bloch_multivector(state1)
plt.show()

print("\nQSphere")
plot_state_qsphere(state1)
plt.show()

print("="*60)
print("Exercise 2 : Hadamard Gate")
print("="*60)

qc2 = QuantumCircuit(1)

qc2.h(0)

print(qc2)

qc2.draw("mpl")
plt.show()

state2 = Statevector.from_instruction(qc2)

print("\nStatevector")
print(state2)

print("\nBloch Sphere")
plot_bloch_multivector(state2)
plt.show()

print("\nQSphere")
plot_state_qsphere(state2)
plt.show()

qc_measure = QuantumCircuit(1,1)

qc_measure.h(0)

qc_measure.measure(0,0)

sim = AerSimulator()

result = sim.run(qc_measure, shots=1024).result()

counts = result.get_counts()

print("\nMeasurement Counts")
print(counts)

plot_histogram(counts)
plt.show()

print("="*60)
print("Exercise 3 : Two Qubits")
print("="*60)

qc3 = QuantumCircuit(2)

qc3.h(0)
qc3.h(1)

print(qc3)

qc3.draw("mpl")
plt.show()

state3 = Statevector.from_instruction(qc3)

print("\nStatevector")
print(state3)

print("\nBloch Sphere")
plot_bloch_multivector(state3)
plt.show()

print("\nQSphere")
plot_state_qsphere(state3)
plt.show()

qc4 = QuantumCircuit(2,2)

qc4.h(0)
qc4.h(1)

qc4.measure([0,1],[0,1])

result = sim.run(qc4, shots=4000).result()

counts = result.get_counts()

print("\nMeasurement Counts")
print(counts)

plot_histogram(counts)
plt.show()

print("\nProbability Amplitudes")

for state, value in state3.to_dict().items():
    probability = abs(value)**2
    print(f"{state} : {probability:.3f}")

print("\nTotal Probability =",
      sum(abs(v)**2 for v in state3.data))
```
